```python
# ## H5: Write a program to calculate and print the average of the three numbers. Round the result to two decimal places.

# তিনটি সংখ্যা নির্ধারণ (বা input নেওয়া)
num1 = 10.5
num2 = 20.25
num3 = 15.8

# গড় হিসাব করা (তিনটি সংখ্যার যোগফলকে ৩ দিয়ে ভাগ)
average = (num1 + num2 + num3) / 3

# দশমিকের পর ২ ঘর পর্যন্ত Round করা
rounded_average = round(average, 2)

# ফলাফল প্রিন্ট করা
print("Numbers:", num1, ",", num2, ",", num3)
print("Raw Average:", average)
print("Rounded Average (2 decimal places):", rounded_average)
```

```python
# ## H6: Create a simple calculator that performs basic arithmetic operations based on user input.

# ইউজার থেকে দুটি সংখ্যা ইনপুট নেওয়া
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# অপারেশনের অপশন প্রদর্শন করা
print("\nSelect Operation:")
print("1. Addition (+)")
print("2. Subtraction (-)")
print("3. Multiplication (*)")
print("4. Division (/)")

# ইউজারের পছন্দ ইনপুট নেওয়া
choice = input("\nEnter choice (1/2/3/4): ")

# শর্ত অনুযায়ী অপারেশন সম্পাদন করা
if choice == '1':
    result = num1 + num2
    print(f"\nResult: {num1} + {num2} = {result}")

elif choice == '2':
    result = num1 - num2
    print(f"\nResult: {num1} - {num2} = {result}")

elif choice == '3':
    result = num1 * num2
    print(f"\nResult: {num1} * {num2} = {result}")

elif choice == '4':
    if num2 != 0:
        result = num1 / num2
        print(f"\nResult: {num1} / {num2} = {result}")
    else:
        print("\nError: Division by zero is not allowed!")

else:
    print("\nInvalid Input! Please select a valid choice (1-4).")
```

```python
# ## H7: Write a program that asks the user to enter a single character and determines whether it is a vowel or a consonant.

# ইউজারের কাছ থেকে ইনপুট নেওয়া
char = input("Enter a single character: ")

# বর্ণটি ইংরেজিতে ছোট হাতের বা বড় হাতের হতে পারে, তাই সুবিধার্থে ছোট হাতের (lowercase) করে নেওয়া হলো
char_lower = char.lower()

# ইনপুটটি সঠিক (একটি মাত্র বর্ণ এবং সেটি Alphabet কিনা) তা পরীক্ষা করা
if len(char_lower) == 1 and char_lower.isalpha():
    # Vowel (a, e, i, o, u) কিনা পরীক্ষা করা
    if char_lower in ['a', 'e', 'i', 'o', 'u']:
        print(f"'{char}' is a Vowel.")
    else:
        print(f"'{char}' is a Consonant.")
else:
    print("Invalid input! Please enter a single alphabetic character.")
```

```python
# ## H8: Create a program that converts temperatures between Fahrenheit and Celsius based on user input.

# অপশন প্রদর্শন করা
print("Temperature Converter")
print("1. Celsius to Fahrenheit")
print("2. Fahrenheit to Celsius")

# ইউজারের পছন্দ ইনপুট নেওয়া
choice = input("\nEnter your choice (1 or 2): ")

# শর্ত অনুযায়ী রূপান্তর করা
if choice == '1':
    celsius = float(input("Enter temperature in Celsius: "))
    fahrenheit = (celsius * 9/5) + 32
    print(f"\n{celsius}°C is equal to {round(fahrenheit, 2)}°F")

elif choice == '2':
    fahrenheit = float(input("Enter temperature in Fahrenheit: "))
    celsius = (fahrenheit - 32) * 5/9
    print(f"\n{fahrenheit}°F is equal to {round(celsius, 2)}°C")

else:
    print("\nInvalid choice! Please select 1 or 2.")
```

```python
# ## H9: Write a program that asks the user to enter three numbers and determines which one is the largest.

# ইউজার থেকে ৩টি সংখ্যা ইনপুট নেওয়া
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

# কোনটি সবচেয়ে বড় তা পরীক্ষা করা
if (num1 >= num2) and (num1 >= num3):
    largest = num1
elif (num2 >= num1) and (num2 >= num3):
    largest = num2
else:
    largest = num3

# ফলাফল প্রদর্শন
print(f"\nThe largest number among {num1}, {num2}, and {num3} is: {largest}")
```

```python
# ## H10: Write a program that prompts the user to input a number N and prints all even numbers between 1 and N using a for loop.

# ইউজার থেকে N এর মান ইনপুট নেওয়া
N = int(input("Enter a positive integer (N): "))

print(f"\nEven numbers between 1 and {N}:")

# range(start, stop, step) ব্যবহার করে 2 থেকে শুরু করে 2 ঘর করে লাফিয়ে চলার নিয়ম
for i in range(2, N + 1, 2):
    print(i, end=" ")
```

```python
# ## H11: Write a program that prompts the user to input a number N and calculates the sum of all numbers from 1 to N that are divisible by 3 or 5.

# ইউজার থেকে N এর মান ইনপুট নেওয়া
N = int(input("Enter a positive integer (N): "))

# যোগফল রাখার জন্য Initial Variable
total_sum = 0

# 1 থেকে N পর্যন্ত লুপ চালানো
for i in range(1, N + 1):
    # সংখ্যাটি 3 অথবা 5 দ্বারা নিঃশেষে বিভাজ্য কিনা পরীক্ষা করা
    if (i % 3 == 0) or (i % 5 == 0):
        total_sum += i  # total_sum = total_sum + i

# ফলাফল প্রদর্শন করা
print(f"\nThe sum of all numbers from 1 to {N} that are divisible by 3 or 5 is: {total_sum}")
```

```python
# ## H12: Write a program that prompts the user to input a number n and prints the multiplication table of n (1 to 10) using a for loop.

# ইউজার থেকে n এর মান ইনপুট নেওয়া
n = int(input("Enter a number to print its multiplication table: "))

print(f"\n--- Multiplication Table of {n} ---")

# 1 থেকে 10 পর্যন্ত লুপ চালানো
for i in range(1, 11):
    result = n * i
    print(f"{n} x {i} = {result}")
```
