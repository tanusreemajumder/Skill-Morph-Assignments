# ** part1: Nampy fandamental**

```python
# Creating NumPy arrays
arr1 = np.array([10, 20, 30, 40, 50])

arr2 = np.array([[100, 200, 300], [400, 500, 600]])

print("1D Array:", arr1)
print("2D Array:\n", arr2)
print("Array shape:", arr1.shape)
print("Array dtype:", arr1.dtype)
```

# **Array creation function zeros,ones,arange,linspace**

```python
# Different ways to create arrays
zeros_array = np.zeros((13, 14), dtype = int)

ones_array = np.ones((22, 33), dtype = int)

range_array = np.arange(0, 100, 22)

linspace_array = np.linspace(0, 10, 50)

print("Zeros array:\n", zeros_array)
print("Ones array:\n", ones_array)
print("Range array:", range_array)
print("Linspace array:", linspace_array)
```

# **Math,Statistic,broadcasting**






```python
# Mathematical operations on arrays
a = np.array([11, 20, 83, 100, 200])
b = np.array([22, 38, 40, 57, 69])

# Element-wise operations
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Power:", a ** 5)
print("Square root:", np.sqrt(a))

# Statistical operations
print("Mean:", np.mean(a))
print("Standard deviation:", np.std(a))
print("Min:", np.min(a))
print("Max:", np.max(a))

# Broadcasting example
matrix = np.array([[11, 12, 13], [14, 15, 16]])
vector = np.array([100, 200, 300])
result = matrix + vector
print("Broadcasting result:\n", result)
```

# **Indexing and slicing arrays**

```python
arr = np.array([101, 201, 301, 401, 501])

# Indexing
print("Element at index 2:", arr[3])

# Slicing
print("Sliced array (from index 1 to 3):", arr[1:4])

# 2D array slicing
arr_2d = np.array([[13, 23, 33], [43, 53, 63], [73, 83, 93]])
print("Sliced 2D array (first two rows):\n", arr_2d[:3])
```

# **Part2: pandas Fundamentals**
creating a dataframe from a dictionary

```python
data = {'Name': ['Alice', 'Bob', 'Charlie', 'Martin', 'Moon', 'Piter'],
        'Age': [24, 27, 22, 29, 32, 21],
        'City': ['New York', 'Los Angeles', 'Chicago', 'Las Vegas', 'Atlanta', 'Miami']}

df = pd.DataFrame(data)
print("DataFrame:\n", df)
```

# **Inspecting a DataFrame — info, describe, head, tail, dtypes**

```python
# Basic DataFrame information
print("\nDataFrame Info:")
print(df.info())
print("\nDataFrame Description:")
print(df.describe())
print("\nDataFrame Shape:", df.shape)
print("Column names:", df.columns.tolist())
# Display first 5 rows
print("First 5 rows of the dataset:")
print(df.head())
# Display last 5 rows
print("\nLast 5 rows of the dataset:")
print(df.tail())
# Check data types
print("\nData Types:")
print(df.dtypes)
```

# **Selecting Single & Multiple Columns**

```python
#Accessing a single column
print("Names column:\n", df['Name'])

# Accessing multiple columns
print("\nName and Age columns:\n", df[['Name', 'Age']])
```

# **Converting a NumPy Array into a DataFrame**

```python
# Convert NumPy array to DataFrame
np_array = np.random.randn(7, 3)
df_from_array = pd.DataFrame(np_array, columns=['A', 'B', 'C'])
print("DataFrame from NumPy array:")
print(df_from_array)
```

# **Grouping & Aggregation — Sales by Region**

```python
# Create sample sales data
sales_data = {
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Product': ['A', 'A', 'B', 'B', 'A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 120, 110, 180, 190, 140],
    'Quantity': [10, 15, 20, 12, 11, 18, 19, 14]
}
sales_df = pd.DataFrame(sales_data)
print("Sales DataFrame:")
print(sales_df)

# Group by operations
region_sales = sales_df.groupby('Region')['Sales'].sum()
print("\nSales by Region:")
print(region_sales)
```
