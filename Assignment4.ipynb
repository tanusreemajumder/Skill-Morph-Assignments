```python
students = pd.read_csv('/content/drive/MyDrive/StudentsPerformance.csv')
gym      = pd.read_csv('/content/drive/MyDrive/gym_membership.csv')

print(students.shape, gym.shape)
```

# **Matplotlib**

```python
# 1. Bar chart of membership types
plt.figure(figsize=(8, 5))
counts = gym['abonoment_type'].value_counts()
plt.bar(counts.index.astype(str), counts.values, color='blue', edgecolor='black')
plt.title('Membership Types Distribution')
plt.xlabel('Abonoment Type')
plt.ylabel('Count')
plt.tight_layout()
plt.show()



```

```python
# 2. Histogram of avg_time_in_gym
plt.figure(figsize=(8, 5))
plt.hist(gym['avg_time_in_gym'], bins=15, color='teal', edgecolor='black')
plt.title('Distribution of Average Time in Gym')
plt.xlabel('Average Time in Gym')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# 3. Scatter plot of Age vs visit_per_week
plt.figure(figsize=(8, 5))
plt.scatter(gym['Age'], gym['visit_per_week'], alpha=0.6, color='coral', edgecolors='k')
plt.title('Age vs Visits per Week')
plt.xlabel('Age')
plt.ylabel('Visits per Week')
plt.tight_layout()
plt.show()
```

# **Seaborn**

```python
# 1. Box plot — gender vs Age
plt.figure(figsize=(8, 5))
sns.boxplot(data=gym, x='gender', y='Age', palette='Set2')
plt.title('Age Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Age')
plt.tight_layout()
plt.show()
```

```python
# 2. Count plot — days_per_week
plt.figure(figsize=(8, 5))
sns.countplot(data=gym, x='days_per_week', palette='rocket')
plt.title('Distribution of Days per Week')
plt.xlabel('Days per Week')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# 3. Bar plot — mean visit_per_week per abonoment_type
plt.figure(figsize=(8, 5))
sns.barplot(data=gym, x='abonoment_type', y='visit_per_week', errorbar=None, palette='mako')
plt.title('Mean Visits per Week by Abonoment Type')
plt.xlabel('Abonoment Type')
plt.ylabel('Mean Visits per Week')
plt.tight_layout()
plt.show()
```
